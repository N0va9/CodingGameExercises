# Formatage pratique

```python
# Pour retirer les .0 inutiles -> :g
x1, y1 = [int(i) for i in input().split()]
x2, y2 = [int(i) for i in input().split()]

print(f"{(x1 + x2)/2:g} {(y1 +y2)/2:g}")

# Avoir l'alphabet rapidement
import string
print(string.ascii_uppercase) #'ABCDE..'
print(string.ascii_lowercase) #'abcde..'
```
