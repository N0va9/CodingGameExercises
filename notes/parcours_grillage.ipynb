# Parcours grillage

Pour parcourir des grilles et la partager en 4 par exemple, il y a 2 options efficaces :

```python
# Avec listes de listes
rows=[]
m, n, k = [int(i) for i in input().split()]
for i in range(m):
    rows.append([int(i) for i in input().split()])
c=0

for i in range(int(m/k)):
    for j in range(int(n/k)):
        t=0
        for a in range(k):
            for b in range(k):
                t+=rows[i*k+a][j*k+b]
        if t>c:
            c=t

print(c)
```
