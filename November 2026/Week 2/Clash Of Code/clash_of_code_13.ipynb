# Clash Of Code 2
Le plus rapide

Mode : Reverse

```python
a = []
e = []

for i in range(7):
    allowance = int(input())
    a.append(allowance)

for i in range(7):
    expense = int(input())
    e.append(expense)

print("Total Allowance:", sum(a) )
print("Total Expenses:", sum(e))
print("Weekly Savings:", sum(a) - sum(e))
```

# Clash Of Code 3
#### Énoncé :
Déterminer si le commerce d'un entrepreneur est rentable ou non

Auteur : mrbisme
Mode : Le plus rapide

```python
fc = int(input())
vc = int(input())
revenue = int(input())

total_cost = fc + vc
net_result = revenue - total_cost

if revenue == 0:
    margin = -100
else:
    margin = (net_result / revenue) * 100

if net_result == 0:
    print("BREAK-EVEN")

else:
    print(f"{ 'PROFITABLE' if net_result > 0 else 'LOSS'} | Net: {net_result} | Margin: {round(margin)}%")
```

# Clash Of Code 4
#### Énoncé :

Permutation

Auteur : ASpookyCombinator

Mode : Le plus rapide

```python
s = input()
res = []

m, n = [int(i) for i in input().split()]
for i in range(m):
    for j in input().split():
        x = int(j)
        res.append(s[x])
    s = ''.join(res)
    res = []

print(s)
```

# Clash Of Code 7
#### Énoncé :
Si nombre pair, ajouter au total sinon soustraire

Auteur : Telokis

Mode : Le plus rapide

```python
total = 0
a = int(input())
b = int(input())
c = int(input())

for e in [a, b, c]:
    if e % 2 == 0:
        total += e
    else:
        total -= e

print(total)
```

# Clash Of Code 8

## Enoncé
Déterminer à quel point on a faim selon ce qu'on a mangé

Auteur : _pongu

Mode : Le plus rapide

```python
def get_score(s):
    return "0ABCDEFGHIJKLMNOPQRSTUVWXYZ".index(s)


n = int(input())
max_points = int(input())
total_points = 0

for i in range(n):
    food = input()
    total_points += get_score(food[0])

if (max_points == 0 and n == 0) or total_points <= 0.2 * max_points:
    print("starving...")

elif 0.2 * max_points < total_points <= 0.5 * max_points:
    print("Yes, please!")

elif 0.5 * max_points < total_points <= 0.85 * max_points:
    print("I wouldn't mind eating something.")

elif 0.85 * max_points < total_points < max_points:
    print("Just a little bit.")

elif max_points <= total_points <= 1.15 * max_points:
    print("No thanks, I'm full.")

else:
    print("BURSTING!!!")
```

# Clash Of Code 9
#### Énoncé :


Auteur : b7241m

Mode : Reverse

```python
res = []

_input = input()

texts = _input.split()

for e in texts:
    if len(e) % 2 != 0:
        res.append(e[::-1])
    else:
        res.append(e)

print(*res)



# Autre version
print(*(w[::(-1)**len(w)] for w in input().split()))
```
