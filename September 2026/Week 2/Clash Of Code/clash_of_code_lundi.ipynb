# Clash Of Code 4
#### Énoncé :

Race engineer

Auteur : PaulBarte

Mode : Le plus rapide

```python
tire_wear = int(input())
laps = int(input())
estimated_life = int(input())
rival_gap = float(input())

if tire_wear > 70:
    print("EMERGENCY_PIT")
elif laps > estimated_life:
    print("SCHEDULED_PIT")
elif rival_gap < 1.5 and tire_wear > 30:
    print("UNDERCUT_DEFENSE")
elif rival_gap  < 5.0 and tire_wear <= 30:
    print("PUSH_HARD")
else:
    print("MAINTAIN_GAP")
```

# Clash Of Code 6
Énoncé :
Racine carré du produit min max
Auteur : MaloDaHood

Mode : Le plus rapide

```python
import math
n = int(input())
if n == 0:
    print(0)
else:
    res = []

    for i in range(n):
        a = int(input())
        res.append(a)

    print(int(math.sqrt(max(res) * min(res) )))
```

# Clash Of Code 14
#### Énoncé :

#### Enoncé:

Exercice du magicien

Auteur : UndersunIO

Mode : Le plus rapide

```python
res = 0
n = int(input())
for i in range(n):
    a = int(input())

    if a < 0:
        res += abs(a)
    else:
        res -= a


print(res)
```
