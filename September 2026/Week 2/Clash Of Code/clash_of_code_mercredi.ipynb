# Clash Of Code 1
#### Énoncé :
Somme de liste

Mode : Reverse

```python
n = int(input())
res = []
for i in range(n):
    x = int(input())
    res.append(x)


print(sum(res))
```

# Clash Of Code 2
#### Énoncé :
Nombre de caractères différents entre 2 chaînes

Mode : Reverse

```python
line1 = input()
line2 = input()

line2 = line2[::-1]
cpt = 0
for i in range(len(line1)):
    if line1[i] != line2[i]:
        cpt +=1
print(cpt)
```

# Clash Of Code 3
#### Énoncé :
Afficher le diplôme d'une personne

Auteur : Chaslain

Mode : Le plus rapide

```python
n = int(input())
res = []

for i in range(n):
    name = input()
    res.append(name)


for i in range(len(res)):
    if len(res[i].split(","))>1:
        print(res[i].split(",")[1].strip())
    else:
        print("N/A")
```

# Clash Of Code 5
#### Énoncé :
Afficher le caractères ou mot le plus fréquent (si égalité choisir le plus long mot)
Afficher le max de répétitions du caractères ou mot le plus fréquent

Auteur : Anonyme

Mode : Le plus rapide

```python
s = input()
result = max((s[i:j] for i in range(len(s)) for j in range(i + 1, len(s) + 1)), key=lambda x: (s.count(x), len(x))) # Key dans max définit sur quoi on compare
print(result)
print(s.count(result))
```
