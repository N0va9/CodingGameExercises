# Clash Of Code 3
#### Énoncé :

Vous travaillez à la NASA et vérifier que la navette atteint Mars

Auteur : The Pro Code

Mode : Le plus rapide

```python
distance = int(input())
time = int(input())
velocity = int(input())
fuel = int(input())
fuel_consumption = int(input())


if velocity * time < distance:
    print('Failure, Not enough time')

elif fuel * fuel_consumption < distance:
    print('Failure, Not enough fuel')

else:
    print("Welcome to Mars")
```

# Clash Of Code 4
#### Énoncé :
Somme de la différence des éléments consécutifs au carré

Auteur : TheProCode

Mode : Le plus rapide

```python
n = int(input())
r = []
for i in input().split():
    a = int(i)
    r.append(a)

res = []

for i in range(len(r) - 1):
    m = max(r[i+1], r[i])
    m2 = min(r[i+1], r[i])
    res.append((m - m2) ** 2 )

print(sum(res))
```

# Clash Of Code 5
#### Énoncé :

UNO game

Auteur : heljii

Mode : Le plus rapide

```python
n = int(input())
e  = 0

bv, bc = '', ''

for i in range(n):
    inputs = input().split()

    v = int(inputs[0])
    c = inputs[1]
    if i == 0:
        bv, bc = v,c
    else:
        if bv != v and bc != c:
            e+=1
        bv = v
        bc = c

print("Correct" if e==0 else f"Incorrect:{e}")
```

# Clash Of Code 6
#### Énoncé :
Changez `clash` par `code` et inversement.
Non trouvé -> `None`

Auteur: [joshua]


```python
n = int(input())
s = input()
res = []

for chaine in s.split():
    if " " not in s:
        if "clash" in chaine:
            res.append(chaine.replace("clash", "code"))
        if "code" in chaine:
            res.append(chaine.replace("code", "clash"))
    else:
        if chaine == "code":
            res.append("clash")
        elif chaine == "clash":
            res.append("code")
        else:
            res.append(chaine)
if (" ".join(res)) == s :
    print("None")
else:
    print(*res)


# Autre version
n = int(input())
s = input()

print(s.replace('clash','c0de').replace('code','clash').replace('c0de','code')if 'clash'in s or 'code'in s else'None')

# Autre version
n = int(input())
s = input()


if any(sub in s for sub in ('clash', 'code')):
    print(s.replace('clash', 'temp').replace('code', 'clash').replace('temp', 'code'))
else:
    print(None)
```

# Clash Of Code 16
#### Énoncé
Exercice de la foudre et du paratonnerre

Auteur : flygon3

Mode : Le plus rapide

```python
n, p, d = [int(i) for i in input().split()]
cpt = 0
for i in range(n):
    x = int(input())
    if p - d <= x <= p + d:
        cpt+=1

print(cpt)
```
